```python
# used chatgpt for filling neighbourhood using longitude and latitude with the help of geopandas
```

```python
import geopandas as gpd 
```

```python
missing = df[df['neighbourhood'].isna()].copy()

points = gpd.GeoDataFrame(
    missing,
    geometry=gpd.points_from_xy(
        missing['long'],
        missing['lat']
    ),
    crs='EPSG:4326'
)
```

```python
url = "https://data.cityofnewyork.us/api/v3/views/9nt8-h7nd/query.geojson?accessType=DOWNLOAD"

neighbourhoods = gpd.read_file(url)
```

```python
neighbourhoods.head()
```

```python
neighbourhoods[['boroname', 'ntaname', 'geometry']].head()
```

```python
neighbourhoods.columns.tolist()
```

```python
matched = gpd.sjoin(
    points,
    neighbourhoods,
    how='left',
    predicate='within'
)
```

```python
matched[['neighbourhood', 'neighbourhood group', 'ntaname', 'boroname']]
```

```python
matched[matched['ntaname'].isna()][
    ['neighbourhood group', 'lat', 'long']
]
```

```python
matched[['neighbourhood group', 'ntaname', 'boroname']]
```

```python
matched[['neighbourhood', 'ntaname']].drop_duplicates()
```

```python
exact_matches = [
    'Park Slope',
    'East Village',
    'Lower East Side',
    'Greenpoint'
]

mask = matched['ntaname'].isin(exact_matches)

df.loc[matched.index[mask], 'neighbourhood'] = matched.loc[mask, 'ntaname']
```

```python
df.loc[
    df['neighbourhood'].isna(),
    ['neighbourhood group', 'lat', 'long']
]
```

```python
neighbourhood_mapping = {
    553: 'Chinatown',
    575: 'Upper West Side',
    624: 'Midtown',
    643: 'Bedford-Stuyvesant',
    726: "Hell's Kitchen",
    731: 'Upper West Side',
    770: 'Sunset Park'
}

df['neighbourhood'] = df['neighbourhood'].fillna(
    pd.Series(neighbourhood_mapping)
)
```
